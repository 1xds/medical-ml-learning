```python
import os
import xml.etree.ElementTree as ET

# 改成你的XML文件夹路径
xml_root = r"C:\Users\Lenovo\Downloads\LIDC-XML-only"

# --- 1. 收集所有XML文件路径 ---
xml_files = []
for root, dirs, files in os.walk(xml_root):
    for f in files:
        if f.lower().endswith('.xml'):
            xml_files.append(os.path.join(root, f))

total = len(xml_files)
print(f"共找到 {total} 个XML文件。")

# --- 2. 解析前50个文件来提取标签（结构一致，没必要全部解析，节省时间）---
# 如果你想解析全部，把 xml_files[:50] 改成 xml_files
sample_files = xml_files  # 全部解析
all_tags = set()
parse_success = 0

print(f"正在解析前 {len(sample_files)} 个文件提取标签...")

for file_path in sample_files:
    try:
        tree = ET.parse(file_path)
        root_elem = tree.getroot()
        # 遍历所有元素，提取标签
        for elem in root_elem.iter():
            all_tags.add(elem.tag)
        parse_success += 1
    except ET.ParseError as e:
        print(f"⚠️ 解析失败: {os.path.basename(file_path)} - {e}")
    except Exception as e:
        print(f"⚠️ 处理失败: {os.path.basename(file_path)} - {e}")

print(f"\n成功解析 {parse_success} 个文件。")
print(f"共提取到 {len(all_tags)} 个唯一标签。\n")

# --- 3. 输出标签列表 ---

# 3.1 简洁版（去掉命名空间）
print("=" * 60)
print("【简洁版标签】（去掉命名空间前缀，按字母排序）")
print("=" * 60)
local_tags = set()
for tag in all_tags:
    # 去掉 {...} 前缀
    if '}' in tag:
        local_name = tag.split('}')[-1]
    else:
        local_name = tag
    local_tags.add(local_name)

for name in sorted(local_tags):
    print(f"  {name}")

# 3.2 完整版（含命名空间）
print("\n" + "=" * 60)
print("【完整版标签】（含命名空间URI，便于定位）")
print("=" * 60)
for tag in sorted(all_tags):
    print(f"  {tag}")

# --- 4. 额外信息：常用关键标签的路径示例 ---
print("\n" + "=" * 60)
print("【常用关键标签说明】")
print("=" * 60)
print("  SeriesInstanceUid     → 患者/序列的唯一标识（你之前提取的UID）")
print("  PatientID             → 患者编号（如 LIDC-IDRI-0001）")
print("  malignancy            → 恶性程度评分（1-5）")
print("  readingSession        → 每一位放射科医生的读片记录")
print("  unblindedReadNodule   → 医生标注的结节")
print("  characteristics       → 结节的详细特征（含malignancy）")
print("  noduleID              → 结节编号")
```

```python
import os
import xml.etree.ElementTree as ET
import pandas as pd


# ============================
# XML路径
# ============================

xml_root = r"C:\Users\Lenovo\Downloads\LIDC-XML-only"


# ============================
# namespace处理
# ============================

def get_tag(element, name):
    """
    匹配去掉namespace后的标签
    """
    if element.tag.endswith(name):
        return element
    return None



# ============================
# 找所有XML
# ============================

xml_files = []

for root, dirs, files in os.walk(xml_root):
    for f in files:
        if f.endswith(".xml"):
            xml_files.append(
                os.path.join(root,f)
            )


print("XML数量:",len(xml_files))



# ============================
# 开始解析
# ============================

records=[]


for xml_path in xml_files:

    try:

        tree = ET.parse(xml_path)

        root = tree.getroot()


        # -----------------------
        # 1. Study UID
        # -----------------------

        study_uid=None
        series_uid=None


        for elem in root.iter():

            tag = elem.tag.split("}")[-1]


            if tag=="StudyInstanceUID":
                study_uid=elem.text


            if tag=="SeriesInstanceUid":
                series_uid=elem.text



        # -----------------------
        # 2. 找结节
        # -----------------------

        for nodule in root.iter():

            tag=nodule.tag.split("}")[-1]


            if tag!="unblindedReadNodule":
                continue



            nodule_id=None
            malignancy=None


            # noduleID

            for child in nodule.iter():

                ctag=child.tag.split("}")[-1]


                if ctag=="noduleID":
                    nodule_id=child.text



                if ctag=="malignancy":
                    try:
                        malignancy=int(child.text)
                    except:
                        malignancy=None



            # 没有malignancy跳过

            if malignancy is None:
                continue



            # -----------------------
            # ROI信息
            # -----------------------

            sop_uid=None

            roi_points=[]


            for child in nodule.iter():

                ctag=child.tag.split("}")[-1]


                if ctag=="imageSOP_UID":
                    sop_uid=child.text


                if ctag=="edgeMap":

                    x=None
                    y=None

                    for p in child:

                        pname=p.tag.split("}")[-1]

                        if pname=="xCoord":
                            x=int(p.text)

                        if pname=="yCoord":
                            y=int(p.text)


                    if x is not None and y is not None:
                        roi_points.append(
                            (x,y)
                        )



            records.append({

                "study_uid":study_uid,

                "series_uid":series_uid,

                "nodule_id":nodule_id,

                "malignancy":malignancy,

                "sop_uid":sop_uid,

                "roi_points":roi_points,

                "xml_path":xml_path

            })


    except Exception as e:

        print(
            "失败:",
            xml_path,
            e
        )



# ============================
# dataframe
# ============================

df_nodules=pd.DataFrame(records)


print("完成")
print(df_nodules.shape)

df_nodules.head()
```

```python
import numpy as np


# 每个study的类别数量
study_profile = (
    df_nodules
    .groupby(["study_uid","malignancy"])
    .size()
    .unstack(fill_value=0)
)


# 保证5列存在
for c in [1,2,3,4,5]:
    if c not in study_profile.columns:
        study_profile[c]=0


study_profile = study_profile[[1,2,3,4,5]]



# 目标
target_nodule = np.array([
    300,   # class1
    500,   # class2
    800,   # class3
    300,   # class4
    200    # class5
])


selected=[]

current=np.zeros(5)



# 随机打乱Study
studies=study_profile.sample(
    frac=1,
    random_state=42
)


for uid,row in studies.iterrows():

    add=row.values


    # 加入后不要明显超过目标
    if np.sum(current+add <= target_nodule):

        selected.append(uid)

        current += add


    if np.all(current >= target_nodule*0.9):
        break



print(
    "Selected studies:",
    len(selected)
)


print(
    "Current nodules:",
    current.astype(int)
)
```
